## Importando datasets

Dado que la mayoría de datasets vienen comprimidos en carpetas .zip, es necesario solo descargarlos, mientra que los archivos en específico del drivedb de phyisionet, al venir archivos solos es conveniente descargarlos todos mediante un código.

```python
import os
import urllib.request
from bs4 import BeautifulSoup
from urllib.error import HTTPError, URLError

# URL base del dataset
base_url = "https://archive.physionet.org/pn3/drivedb/"
dataset_dir = "Datasets_descargados/drivedb"

# Crear carpeta si no existe
os.makedirs(dataset_dir, exist_ok=True)

# Leer el contenido HTML de la página
print("Obteniendo lista de archivos...")
response = urllib.request.urlopen(base_url)
html = response.read()

# Analizar con BeautifulSoup
soup = BeautifulSoup(html, 'html.parser')

# Buscar todos los enlaces válidos a archivos
valid_exts = ('.dat', '.hea', '.xyz', '.txt', '.xml', '.csv', '.wfdb', '.pdf')
links = [
    a['href'] for a in soup.find_all('a', href=True)
    if a['href'].lower().endswith(valid_exts)
]

print(f"Se encontraron {len(links)} archivos para descargar.\n")

# Descargar cada archivo si no existe localmente
for file_name in links:
    file_url = base_url + file_name
    local_path = os.path.join(dataset_dir, file_name)
    
    if os.path.exists(local_path):
        print(f"{file_name} ya existe, se omite.")
        continue

    try:
        print(f"Descargando {file_name}...")
        urllib.request.urlretrieve(file_url, local_path)
    except (HTTPError, URLError) as e:
        print(f"⚠️ No se pudo descargar {file_name}: {e}")

print("\n✅ Descarga completa. Archivos guardados en:", dataset_dir)
```

```python
# === PARTE 2: LECTURA DE ETIQUETAS DESDE ARCHIVOS .hea ===
print("\nAnalizando etiquetas de los archivos .hea...\n")

hea_files = [f for f in os.listdir(dataset_dir) if f.endswith(".hea")]

if not hea_files:
    print("No se encontraron archivos .hea en el dataset.")
else:
    for hea in hea_files:
        path = os.path.join(dataset_dir, hea)
        with open(path, "r") as f:
            lines = f.readlines()

        print(f"📄 Archivo: {hea}")
        # La primera línea suele contener: nombre, número de señales, frecuencia, etc.
        print("→ Cabecera principal:", lines[0].strip())

        # Cada línea siguiente describe un canal
        print("→ Canales:")
        for line in lines[1:]:
            parts = line.strip().split()
            if len(parts) >= 9:
                channel_label = parts[-1]
                print(f"   - {channel_label}")
        print("-" * 50)
```
